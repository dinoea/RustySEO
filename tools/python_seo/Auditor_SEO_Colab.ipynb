Este notebook ejecuta el crawler `auditor_seo_es.py`, genera CSV + HTML y permite descargar los resultados.

```python
from pathlib import Path
from google.colab import files

# Si montas tu repo en Drive, actualiza esta ruta
RUTA_SCRIPT = Path('/content/RustySEO/tools/python_seo/auditor_seo_es.py')
assert RUTA_SCRIPT.exists(), f'No se encontró: {RUTA_SCRIPT}'
```

```python
import importlib.util

spec = importlib.util.spec_from_file_location('auditor_seo_es', RUTA_SCRIPT)
auditor = importlib.util.module_from_spec(spec)
spec.loader.exec_module(auditor)
```

```python
# Configuración
DOMINIO = 'https://example.com'
MAX_PAGINAS = 25
DELAY = 0.2
OUT_DIR = '/content/salida_auditoria'
CSV_NAME = 'auditoria_seo.csv'

resultados, csv_path = auditor.ejecutar_auditoria(
    dominio=DOMINIO,
    max_paginas=MAX_PAGINAS,
    delay=DELAY,
    output_dir=OUT_DIR,
    csv_name=CSV_NAME,
)

print(f'URLs analizadas: {len(resultados)}')
print(f'CSV: {csv_path}')
print(f'HTML dir: {Path(OUT_DIR) / "html_exportado"}')
```

```python
# Descargar CSV
files.download(str(csv_path))
```

```python
# Comprimir carpeta de HTML y descargar
import shutil
from pathlib import Path

zip_base = '/content/html_exportado'
html_dir = Path(OUT_DIR) / 'html_exportado'
zip_file = shutil.make_archive(zip_base, 'zip', html_dir)
files.download(zip_file)
```
